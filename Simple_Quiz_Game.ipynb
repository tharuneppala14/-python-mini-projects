```python
# Simple Quiz Game

print("===== SIMPLE QUIZ =====")

score = 0

answer = input("1. What is the capital of India? ")

if answer.lower() == "new delhi":
    print("Correct!")
    score = score + 1
else:
    print("Wrong!")

answer = input("2. Which language are we learning? ")

if answer.lower() == "python":
    print("Correct!")
    score = score + 1
else:
    print("Wrong!")

answer = input("3. How many days are there in a week? ")

if answer == "7":
    print("Correct!")
    score = score + 1
else:
    print("Wrong!")

print("\n===== RESULT =====")
print("Your score is:", score, "/ 3")

if score == 3:
    print("Excellent!")
elif score == 2:
    print("Good job!")
else:
    print("Keep practicing!")
```
