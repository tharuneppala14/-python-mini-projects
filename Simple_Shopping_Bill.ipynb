```python
# Simple Shopping Bill

print("===== SHOPPING MENU =====")
print("1. Notebook - ₹50")
print("2. Pen      - ₹20")
print("3. Pencil   - ₹10")
print("4. Bag      - ₹500")

choice = int(input("Enter your choice: "))
quantity = int(input("Enter quantity: "))

if choice == 1:
    item = "Notebook"
    price = 50

elif choice == 2:
    item = "Pen"
    price = 20

elif choice == 3:
    item = "Pencil"
    price = 10

elif choice == 4:
    item = "Bag"
    price = 500

else:
    print("Invalid choice!")
    exit()

total = price * quantity

print("\n===== BILL =====")
print("Item:", item)
print("Price: ₹", price)
print("Quantity:", quantity)
print("Total Amount: ₹", total)

print("Thank you for shopping!")
```
