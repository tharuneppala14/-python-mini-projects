```python
# Student Grade Calculator

print("===== STUDENT GRADE CALCULATOR =====")

name = input("Enter student name: ")
marks = float(input("Enter marks: "))

if marks >= 90:
    grade = "A"
elif marks >= 80:
    grade = "B"
elif marks >= 70:
    grade = "C"
elif marks >= 60:
    grade = "D"
else:
    grade = "F"

print("\n===== RESULT =====")
print("Student Name:", name)
print("Marks:", marks)
print("Grade:", grade)

if grade == "F":
    print("Result: Fail")
else:
    print("Result: Pass")
```
